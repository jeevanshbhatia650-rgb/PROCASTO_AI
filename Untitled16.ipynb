```python
import os
import pandas as pd
import chromadb
from fastembed import TextEmbedding # Import TextEmbedding directly from fastembed

# Define a custom embedding function class to wrap fastembed.TextEmbedding
class CustomFastEmbedEmbeddingFunction:
    def __init__(self, model_name: str = "BAAI/bge-small-en-v1.5"):
        self.model = TextEmbedding(model_name)
        self._model_name = model_name # Store model name for the 'name' property

    def __call__(self, input: list[str]) -> list[list[float]]: # Changed 'texts' to 'input'
        # FastEmbed's embed method returns a list of np.ndarray
        embeddings_np = self.model.embed(input)
        # Convert numpy arrays to lists of floats as expected by ChromaDB
        return [emb.tolist() for emb in embeddings_np]

    def name(self):
        return self._model_name # Change name to a method as expected by ChromaDB

# 1. Initialize persistent ChromaDB storage in Colab
db_path = "/content/chroma_data"
chroma_client = chromadb.PersistentClient(path=db_path)

# 2. Configure embedding function using the custom wrapper
embedding_func = CustomFastEmbedEmbeddingFunction(model_name="BAAI/bge-small-en-v1.5")

collection = chroma_client.get_or_create_collection(
    name="samsung_appliance_codes",
    embedding_function=embedding_func,
    metadata={"hnsw:space": "cosine"}
)

# 3. Load dataset
csv_path = "/content/samsung_appliance_codes.csv"
df = pd.read_csv(csv_path).fillna("")

# 4. Convert records to natural language
def build_natural_language_doc(row):
    appliance = str(row['appliance_type']).strip().capitalize()
    code = str(row['code']).strip()
    meaning = str(row['meaning']).strip().rstrip('.')

    doc_lines = [
        f"Samsung {appliance} error code '{code}' indicates: {meaning}."
    ]

    details = []
    if row['component']:
        details.append(f"affected component is the {row['component']}")
    if row['cause_category']:
        details.append(f"cause category relates to {row['cause_category']}")
    if row['severity']:
        details.append(f"severity is rated as {row['severity']}")
    if row['diy_difficulty']:
        details.append(f"repair difficulty is {row['diy_difficulty']}")

    if details:
        doc_lines.append(f"System diagnostics state that the {', and the '.join(details)}.")

    if row['fix_steps']:
        clean_steps = str(row['fix_steps']).strip()
        doc_lines.append(f"Troubleshooting and repair steps: {clean_steps}")

    return " ".join(doc_lines)

documents = []
metadatas = []
ids = []

for idx, row in df.iterrows():
    documents.append(build_natural_language_doc(row))
    metadatas.append({
        "brand": str(row['brand']),
        "market": str(row['market']),
        "appliance_type": str(row['appliance_type']).lower(),
        "code": str(row['code']).upper(),
        "severity": str(row['severity']),
        "diy_difficulty": str(row['diy_difficulty']),
        "component": str(row['component'])
    })
    ids.append(f"{row['appliance_type']}_{row['market']}_{row['code']}_{idx}")

# 5. Insert records into ChromaDB
batch_size = 50
for i in range(0, len(documents), batch_size):
    collection.add(
        documents=documents[i:i+batch_size],
        metadatas=metadatas[i:i+batch_size],
        ids=ids[i:i+batch_size]
    )

print(f"Embedding complete! Stored {collection.count()} fault codes in ChromaDB.")
```
